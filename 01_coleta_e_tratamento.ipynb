```python
import pandas as pd
import numpy as np
import requests
```

```python
url_periodo1 = 'https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados?formato=json&dataInicial=01/07/1994&dataFinal=01/07/2004'
```

```python
url_periodo2 = 'https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados?formato=json&dataInicial=02/07/2004&dataFinal=01/07/2014'
```

```python
url_periodo3 = 'https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados?formato=json&dataInicial=02/07/2014&dataFinal=01/07/2024'
```

```python
url_periodo4 = 'https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados?formato=json&dataInicial=02/07/2024&dataFinal=19/10/2026'
```

```python
dados1 = requests.get(url_periodo1).json()
```

```python
dados2 = requests.get(url_periodo2).json()
```

```python
dados3 = requests.get(url_periodo3).json()
```

```python
dados4 = requests.get(url_periodo4).json()
```

```python
df1 = pd.DataFrame(dados1)
```

```python
df2 = pd.DataFrame(dados2)
```

```python
df3 = pd.DataFrame(dados3)
```

```python
df4 = pd.DataFrame(dados4)
```

```python
df = pd.concat([df1, df2, df3, df4])
```
